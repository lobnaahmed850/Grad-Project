```python
#handling missing/inf values
import numpy as np

# ---- Check before fixing ----
print("Missing values per column (only showing columns with >0):")
missing = df_plus.isnull().sum()
print(missing[missing > 0])

numeric_cols = df_plus.select_dtypes(include=[np.number]).columns
inf_counts = np.isinf(df_plus[numeric_cols]).sum()
print("\nInf values per column (only showing columns with >0):")
print(inf_counts[inf_counts > 0])

# ---- Fix: replace inf/-inf with NaN first, so we can handle them the same way as missing values ----
df_plus[numeric_cols] = df_plus[numeric_cols].replace([np.inf, -np.inf], np.nan)

# ---- Decide how to handle remaining NaNs ----
# Option A (recommended if NaN rows are a small % of total): drop the rows
nan_row_pct = df_plus.isnull().any(axis=1).mean() * 100
print(f"\nRows with at least one NaN: {nan_row_pct:.3f}% of dataset")

if nan_row_pct < 1:  # adjust this threshold based on what you see
    df_plus = df_plus.dropna().reset_index(drop=True)
    print(f"Dropped NaN rows. New shape: {df_plus.shape}")
else:
    # Option B (if NaN rows are a larger %, impute instead of losing too much data)
    for col in numeric_cols:
        if df_plus[col].isnull().sum() > 0:
            median_val = df_plus[col].median()
            df_plus[col] = df_plus[col].fillna(median_val)
    print(f"Imputed NaNs with column medians. New shape: {df_plus.shape}")

# ---- Final verification ----
print("\nRemaining missing values:", df_plus.isnull().sum().sum())
print("Remaining inf values:", np.isinf(df_plus[numeric_cols]).sum().sum())
```

```python
#----Drop identifier columns (Flow ID, IPs, Timestamp)----
# These are columns that identify a specific flow instance rather than
# describing its traffic characteristics -- they don't generalize and
# can cause data leakage or errors in scaling/model fitting.

identifier_cols = [
    'flow id',
    'source ip',
    'destination ip',
    'timestamp',
    'src ip',          # alternate naming some versions use
    'dst ip',          # alternate naming some versions use
]

# only drop columns that actually exist in this df (names vary by dataset version)
cols_to_drop = [col for col in identifier_cols if col in df_plus.columns]
print(f"Dropping these identifier columns: {cols_to_drop}")

df_plus = df_plus.drop(columns=cols_to_drop)
print(f"Shape after dropping identifiers: {df_plus.shape}")
```
